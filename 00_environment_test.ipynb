```python
# ============================================
# Data Science Environment Final Test
# ============================================

import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# -------------------------
# 1. NumPy
# -------------------------

numbers = np.array([10, 20, 30, 40, 50])

print("NumPy Mean:", numbers.mean())
print("NumPy Maximum:", numbers.max())


# -------------------------
# 2. Pandas
# -------------------------

data = {
    "Hours": [1, 2, 3, 4, 5],
    "Score": [20, 40, 60, 80, 100]
}

df = pd.DataFrame(data)

print("\nPandas DataFrame:")
display(df)


# -------------------------
# 3. Scikit-learn
# -------------------------

X = df[["Hours"]]
y = df["Score"]

model = LinearRegression()
model.fit(X, y)

prediction = model.predict([[6]])

print("\nScikit-learn Prediction:")
print("Score for 6 hours:", prediction[0])


# -------------------------
# 4. Matplotlib
# -------------------------

plt.figure(figsize=(8, 5))

plt.scatter(df["Hours"], df["Score"])
plt.plot(df["Hours"], model.predict(X))

plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.title("Study Hours vs Score")

plt.show()


print("\n================================")
print("DATA SCIENCE ENVIRONMENT: READY")
print("================================")
```
